H. Python Operators Usecases* *Use Case 1: Internet Data Usage Calculator

```python
limit = 10
used = 8

print(int(limit))
print(int(used))

Remainingdata = limit - used
Usage_percentage = (used / limit) * 100

print(Remainingdata, "GB")
print(Usage_percentage, "%")

if Usage_percentage >= 80:
    print("Warning: High usage, consider upgrading your plan.")
```

Use Case 2: Shopping Discount Calculation

```python
price = 1000.67
Discount_percent = 20

print(float(price))
print(int(Discount_percent))


Discountamount = (price * Discount_percent) / 100
Finalprice = price - Discountamount

print("Original price=", price)
print("discount applied=" ,Discountamount)
print("final payable amount=", Finalprice)
```
